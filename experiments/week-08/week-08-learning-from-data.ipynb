## 1. Set up

The first cell loads the libraries. `numpy` holds the data, `matplotlib` draws the plots, and `scikit-learn` (imported as `sklearn`) provides the learning algorithms.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.neighbors import KNeighborsClassifier
from sklearn.cluster import KMeans
```

If this cell fails with `ModuleNotFoundError`, the libraries are not installed in this environment. That does not happen in Colab. In VS Code, install them as described above, then run the cell again.

## 2. The data

We use six examples. Each example has two numerical features, so each one can be drawn as a point. In machine-learning code the features are usually called `X`, with one row per example. The labels are called `y`, with one label per example.

`unknown` is a new example. It is not one of the six, and it has no label.

```python
X = np.array([[1.0, 1.0], [1.5, 2.0], [2.0, 1.3],
              [5.8, 6.4], [6.0, 5.8], [6.5, 5.5]])
y = np.array(["A", "A", "A", "B", "B", "B"])
unknown = np.array([[2.4, 1.8]])

print("X has shape", X.shape)
print("y has shape", y.shape)
```

`X.shape` is `(6, 2)`. That means six examples with two features each. Checking shapes is a good habit, because many errors in machine-learning code come from data with the wrong shape.

## 3. Look at the data

This cell defines a small plotting function, which the rest of the notebook reuses. Then it draws the six labelled examples and the unknown point. Later plots use different colours on purpose, so that groups found without labels never look like the classes A and B.

```python
def plot_points(X, groups, title, unknown=None, unknown_label="?", first_colour=0):
    fig, ax = plt.subplots(figsize=(5, 5))
    for i, group in enumerate(sorted(set(groups))):
        mask = groups == group
        ax.scatter(X[mask, 0], X[mask, 1], s=90, color=f"C{i + first_colour}", label=str(group))
    if unknown is not None:
        ax.scatter(unknown[:, 0], unknown[:, 1], s=160, marker="D",
                   facecolors="white", edgecolors="black", linewidths=2)
        ax.annotate(unknown_label, unknown[0], xytext=(8, 8),
                    textcoords="offset points", fontsize=12, weight="bold")
    ax.set(xlim=(0, 8), ylim=(0, 8), xlabel="Feature 1", ylabel="Feature 2", title=title)
    ax.set_aspect("equal")
    ax.grid(alpha=0.3)
    ax.legend(loc="upper left")
    plt.show()

plot_points(X, y, "Six labelled examples and one unknown point", unknown)
```

**Predict before you run the next cell.** Is the unknown point A or B? What in the picture makes you think so?

## 4. Classification

Now let a classifier decide. `KNeighborsClassifier` is k-nearest neighbours, which Week 9 covers in detail.

The code follows the same pattern for almost every model in scikit-learn. You create the model, you `fit` it to labelled examples, and you `predict` for new ones.

```python
model = KNeighborsClassifier(n_neighbors=3)
model.fit(X, y)
prediction = model.predict(unknown)

print("Predicted class for the unknown point:", prediction[0])
```

```python
plot_points(X, y, "The prediction", unknown, unknown_label=prediction[0])
```

The model gave the new point one of the classes it was shown, A or B. That is what classification does. One prediction does not tell us how well the model works on other points.

## 5. Clustering

Now remove the labels. `X` stays exactly as it is, and we simply do not use `y`. The unknown point is left out, because it was never one of the six examples.

```python
no_labels = np.array(["no label"] * len(X))
plot_points(X, no_labels, "The same six examples, without labels", first_colour=7)
```

**Predict before you run the next cell.** If you had to split these points into two groups, which points would go together?

```python
model = KMeans(n_clusters=2, random_state=42, n_init=10)
groups = model.fit_predict(X)

print("Group for each example:", groups)
plot_points(X, groups, "Groups found without labels", first_colour=2)
```

Notice that we asked for two clusters. K-means did not work out how many groups there are. We told it.

The group numbers, 0 and 1, are only names. They are not A and B, and swapping them would change nothing. `random_state` and `n_init` fix the starting conditions, so the result is the same every time you run the cell. Week 10 covers how K-means works.

Compare the two `fit` calls. The classifier received `X` and `y`. K-means received only `X`. In code, that is the difference between supervised and unsupervised learning.
